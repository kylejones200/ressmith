# PetroSmith Tutorial 1: Basic Drilling Calculations

This notebook demonstrates fundamental drilling calculations using the PetroSmith library.

## Topics Covered
- Hydrostatic pressure
- Equivalent Circulating Density (ECD)
- Casing design
- Error handling

```python
# Import the library
import sys
sys.path.insert(0, '..')

from petrosmith.api import DrillingAPI
from petrosmith.exceptions import InvalidMudWeightError
```

## 1. Hydrostatic Pressure

Calculate the pressure exerted by a column of drilling mud.

```python
# Initialize API
api = DrillingAPI()

# Calculate hydrostatic pressure
mud_weight = 10.5  # ppg
tvd = 10000  # feet

pressure = api.calculate_hydrostatic_pressure(mud_weight, tvd)

print(f"Mud Weight: {mud_weight} ppg")
print(f"True Vertical Depth: {tvd:,} ft")
print(f"Hydrostatic Pressure: {pressure:,.0f} psi")
```

## 2. Comparing Different Mud Weights

Let's compare pressures at different depths and mud weights.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Create depth range
depths = range(0, 15001, 1000)
mud_weights = [9.0, 10.5, 12.0, 14.0]

# Calculate pressures
results = []
for mw in mud_weights:
    pressures = [api.calculate_hydrostatic_pressure(mw, d) for d in depths]
    results.append(pressures)

# Plot
plt.figure(figsize=(10, 6))
for i, mw in enumerate(mud_weights):
    plt.plot(results[i], depths, label=f"{mw} ppg")

plt.xlabel('Pressure (psi)')
plt.ylabel('Depth (ft)')
plt.title('Hydrostatic Pressure vs Depth')
plt.legend()
plt.grid(True)
plt.gca().invert_yaxis()  # Invert y-axis (depth increases downward)
plt.show()
```

## 3. Equivalent Circulating Density (ECD)

Calculate ECD when circulating.

```python
# Calculate ECD
annular_pressure_loss = 450  # psi

ecd = api.calculate_ecd(mud_weight, annular_pressure_loss, tvd)

print(f"Static Mud Weight: {mud_weight} ppg")
print(f"Annular Pressure Loss: {annular_pressure_loss} psi")
print(f"ECD: {ecd:.2f} ppg")
print(f"ECD Increase: {ecd - mud_weight:.2f} ppg")
```
